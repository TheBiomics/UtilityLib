# StepManager

```python
from UtilityLib.lib.step import StepManager
```

```python
# Scenario 1: List of stages
sm1 = StepManager(['Stage-1', 'Stage-2', 'Stage-3', 'Stage-4'])
print("List input:")
print(f"Stages: {sm1.stages_list}")
print(f"Priorities: {sm1.priorities}")
print(f"Current: {sm1.current}")
print(f"Is first: {sm1.is_first}, Is last: {sm1.is_last}")
print()
```

```python
# Scenario 5: Test navigation and cycling
print("Testing navigation on sm1:")
print(f"Start: {sm1.current}")
sm1.next_stage()
print(f"After next: {sm1.current}")
sm1.next_stage()
print(f"After next: {sm1.current}")
sm1.next_stage()
print(f"After next: {sm1.current}")
```

```python
# Scenario 2: List of tuples (stage, priority)
sm2 = StepManager([('Stage-1', 1), ('Stage-2', 5), ('Stage-3', 0), ('Stage-4', 3)])
print("List of tuples input:")
print(f"Stages: {sm2.stages_list}")
print(f"Priorities: {sm2.priorities}")
print(f"Current: {sm2.current}")
print(f"Is first: {sm2.is_first}, Is last: {sm2.is_last}")
print()
```

```python
# Scenario 3: Dict (stage: priority)
sm3 = StepManager({'Stage-1': 2, 'Stage-2': 5, 'Stage-3': 0, 'Stage-4': 1})
print("Dict input:")
print(f"Stages: {sm3.stages_list}")
print(f"Priorities: {sm3.priorities}")
print(f"Current: {sm3.current}")
print(f"Is first: {sm3.is_first}, Is last: {sm3.is_last}")
print()
```

```python
# Scenario 4: Positional args + kwargs
sm4 = StepManager('Stage-1', 'Stage-2', Stage_3=10, Stage_4=5)
print("Positional + kwargs input:")
print(f"Stages: {sm4.stages_list}")
print(f"Priorities: {sm4.priorities}")
print(f"Current: {sm4.current}")
print(f"Is first: {sm4.is_first}, Is last: {sm4.is_last}")
print()
```
